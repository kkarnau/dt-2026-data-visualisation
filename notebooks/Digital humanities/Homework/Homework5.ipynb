```python
import pandas as pd 
df = pd.read_csv("Estonian writers.csv")
df
```

In this chart I find the average number on works in each genre. It shows that most works are in prose genre and the least in young adult.

```python
import matplotlib.pyplot as plt
avg_age= df.groupby("Genre")["Works written"].mean()
avg_age.plot(kind="bar")
plt.title("Average number of works in each genre")
plt.ylabel("Number of works")
plt.xlabel("Genre")
plt.show()
```

Correlation between number of works written and number of awards received by writers. Based on the chart it is possible to say that more works equals more awards for the writer. On the other hand, the data varies quite a lot.

```python
plt.scatter(df["Works written"], df["Awards"], s=df["Awards"]*40, alpha=0.4, color="#FFC0CB")
plt.title("Awards vs works written")
plt.ylabel("Awards")
plt.xlabel("Works")
```
