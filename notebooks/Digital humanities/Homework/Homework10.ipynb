I used the same raw text file to create another wordcloud. It is easier to understand, beacuse the background is now white and I removed most of the stopwords. I found a raw text file from Github that contained the most popular stopwords in Estonian. I did not find a good way to remove all the stopwords, which is way some are still quite visible and disturb the whole picture. Now it is clear that the text is about speech and focuses on the syllables.

```python
stopwords_list = urllib.request.urlopen("https://raw.githubusercontent.com/stopwords-iso/stopwords-et/refs/heads/master/raw/gh-stopwords-json-et.txt").read().decode("utf-8").split("\n")
WordCloud(stopwords=stopwords_list, width= 500, height= 300, background_color= "white").generate_from_text(text).to_image()
```

I also tried to make a wordcloud that shows the words in different colours based on the word's length. I used the same text. Words that have less than 5 letters are in blue and longer words are in red. The code has a loop and conditions to check all the words in text.

```python
def color_function(word, font_size, position, orientation, random_state=None, **additional_args):
  if len(word)>4:
    return "red"
  return "blue"

WordCloud().generate_from_text(text) \
  .recolor(color_func=color_function) \
  .to_image()
```

For the last wordcloud I tried to import a picture to shape the wordcloud. I found an okay shape of quatrefoil and imported the picture file from my Google Drive. Again I used the same text about syllables and removed the most popular stopwords. Unfortunately the wordcloud is not in the shape of the quatrefoil but around the outline. I tried to change the code a few times, but I do not know how to fix it. So the words make a shape of a quatrefoil. I used the code to make it pink, adjust the size and font size.

```python
import numpy as np
import urllib.request
from PIL import Image
from wordcloud import WordCloud
from IPython.display import display

file_id = "1TRf_MrctO0k_ATC1ekBhA70D_W6F2v50"

url = f"https://drive.google.com/uc?export=download&id={file_id}"
urllib.request.urlretrieve(url, "Kujund.png")

stopwords_list = urllib.request.urlopen("https://raw.githubusercontent.com/stopwords-iso/stopwords-et/refs/heads/master/raw/gh-stopwords-json-et.txt").read().decode("utf-8").split("\n")

img = Image.open("Kujund.png").convert("L")
img = img.resize((500, 500))

mask = 255 - np.array(img)

wordcloud = WordCloud(stopwords= stopwords_list,
    width=500,
    height=500,
    mask=mask,
    background_color="pink",
    contour_width=0,
    max_words=200,
    min_font_size=5,
    collocations=False
).generate_from_text(text)

display(wordcloud.to_image())
```
