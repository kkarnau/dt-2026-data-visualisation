```python
# can use to work with tabular data
import pandas as pd 
df = pd.read_csv("competition.csv")
df.head() # Show the first few rows
#df.describe() #Show some basic statistics
```

```python
import matplotlib.pyplot as plt #matplotlib helps to creates chart and visualizations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per Athlete")
plt.ylabel("Years of practice")
#plt.show()
```

```python
import matplotlib.pyplot as plt #matplotlib helps to creates chart and visualizations
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age per Athlete")
plt.ylabel("Age")
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Age of Athletes")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6)
#s=df["Years_of_practice"]*100 means if student has more years of practice their bubble is bigger
#alpha=0.6 makes the bubbles more transparent
plt.title("Years of practice based on athlete's age")
plt.ylabel("Years of practice")
plt.xlabel("Age")
```

```python
import seaborn as sns #use this to create more clean and formatted plots
sns.boxplot(data=df, y="Age")
plt.title("Distribution of athlete age")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.show
```

```python
#avg age per each sport. 
avg_age= df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.show()
```

```python
#avg years of practice each sport
avg_age= df.groupby("Sport")["Years_of_practice"].mean()
avg_age.plot(kind="bar")
plt.show()
```
