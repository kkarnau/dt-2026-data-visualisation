```python
import pandas as pd 
df = pd.read_csv("Competencies.csv")
df
#import dataset
```

```python
import matplotlib.pyplot as plt #matplotlib helps to creates chart and visualizations
#create a chart showing who puts in the most effort, average ja värki
#algne idee, hiljem kompktsem #df.plot(kind="bar", x="Student", y="Hours_Studied", legend=False)
plt.title("Amount of effort per student")
plt.ylabel("Hours studied")
avg_age= df.groupby("Student")["Hours_Studied"].mean()
avg_age.plot(kind="bar", x="Student", y="Hours_Studied", color="#FFC0CB")
#colors= plt.cm.plasma(range(len(avg_hours)))
plt.show()
```

```python
plt.scatter(df["Grade"], df["Hours_Studied"], s=df["Hours_Studied"]*40, alpha=0.4, color="#FFC0CB")
#s=df["Years_of_practice"]*100 means if student has more years of practice their bubble is bigger
#alpha=0.6 makes the bubbles more transparent
#plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*20, alpha=0.2)
plt.title("Grades per effort")
plt.ylabel("Hours Studied")
plt.xlabel("Grade")
```
